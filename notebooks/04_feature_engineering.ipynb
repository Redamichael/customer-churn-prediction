This notebook contains structured analysis steps as designed in the project.

```python
# 04_feature_engineering.ipynb

import pandas as pd

df = pd.read_csv("../data/processed/cleaned_churn.csv")

# New features
df['IsNewCustomer'] = (df['tenure'] < 12).astype(int)

df['AvgMonthlySpend'] = df['TotalCharges'] / (df['tenure'] + 1)

df['HighValueCustomer'] = (df['MonthlyCharges'] > df['MonthlyCharges'].median()).astype(int)

# Save enhanced dataset
df.to_csv("../data/processed/featured_churn.csv", index=False)

print("Feature engineering completed.")
```

### Feature Engineering Strategy
#### This notebook transforms cleaned data into predictive signals by engineering domain-specific features that capture customer behavior and value.

* **New Customer Identification:** Flags customers in their first year of service to target high-risk early churn periods.

* **Spend Normalization:** Calculates AvgMonthlySpend with zero-division protection to provide a stable metric for customer investment.

* **Value Segmentation:** Categorizes customers as HighValueCustomer based on median monthly spend to help the model distinguish between different service tiers.

* **Data Persistence:** Exports the enhanced dataset to featured_churn.csv to maintain a modular and reproducible modeling pipeline
