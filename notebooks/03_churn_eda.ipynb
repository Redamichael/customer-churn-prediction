```python
# 03_eda.ipynb

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/cleaned_churn.csv")

# Tenure vs churn
sns.boxplot(x='Churn', y='tenure', data=df)
plt.title("Tenure vs Churn")
plt.show()

# Contract impact
sns.countplot(x='Contract', hue='Churn', data=df)
plt.xticks(rotation=30)
plt.title("Churn by Contract Type")
plt.show()

# Monthly charges
sns.histplot(data=df, x='MonthlyCharges', hue='Churn', bins=30, kde=True)
plt.title("Monthly Charges Distribution")
plt.show()
```

### Findings
* **Tenure vs Churn:** Customers who churn typically have a much shorter tenure, with a median of approximately 10 months compared to nearly 40 months for those who stay.

* **Churn by Contract Type:** Month-to-month contracts experience significantly higher churn rates, while one-year and two-year contracts show very high retention levels.

* **Monthly Charges Distribution:** Customers with higher monthly charges (between 70 and 110) appear to churn more frequently compared to those on low-cost plans around the 20 mark.
