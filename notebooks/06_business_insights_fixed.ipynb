# 06 Business Insights Notebook
Customer segmentation & risk analysis


```python
import pandas as pd
import joblib
```

```python
# Load data and model
df = pd.read_csv('../data/processed/featured_churn.csv')
pipeline = joblib.load('../outputs/model/churn_pipeline.pkl')

X = df.drop('Churn', axis=1)
```

```python
# Predict probabilities
df['Churn_Prob'] = pipeline.predict_proba(X)[:,1]
```

```python
# Risk segmentation
df['Risk'] = pd.cut(
    df['Churn_Prob'],
    bins=[0, 0.3, 0.6, 1],
    labels=['Low','Medium','High']
)
```

```python
# Summary
print(df['Risk'].value_counts())
```

```python
# High-risk customers
high_risk = df[df['Risk'] == 'High']
print('High-risk customers:', len(high_risk))

high_risk.head()
```
